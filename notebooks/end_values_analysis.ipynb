# Notebook to generate a table with the end values of all the different experiments

### Reading and combining in experiment results data

```python
base_and_climate_runs = pd.read_csv('../results/base_and_climate.csv')
policy_runs = pd.read_csv('../results/policies.csv')
policy_comb_runs = pd.read_csv('../results/policies_comb.csv')
```

```python
all_results = pd.concat([base_and_climate_runs, policy_runs, policy_comb_runs], ignore_index=True)
```

```python
all_results.head()
```

### Dataframe with relevant data
- last step
- experiment input parameters
- fraction adapted and damage values

```python
# make df last step
all_results_last_step = all_results[all_results['Step'] == 50]
all_results_last_step.shape
```

```python
# only keep columns of interest
all_results_last_step_red = all_results_last_step[['set_hh_adapted_at_start', 'flood_time', 'flood_depth_in_m','policies_file_path', 'policy_activation','cumulative_damage', 'cumulative_damage_no_adaptation', 'fraction_adapted_any']]
```

```python
# calcualte damage avoided
all_results_last_step_red['damage_avoided'] = (all_results_last_step_red['cumulative_damage_no_adaptation']-all_results_last_step_red['cumulative_damage'])
```

```python
all_results_last_step_red.head()
```

```python
# make shortened policy column
all_results_last_step_red['policy'] = all_results_last_step_red.apply(lambda row: 'None' if (isinstance(row['policies_file_path'], str) and row['policies_file_path'].endswith('policies_null.json')) or str(row['policy_activation']).strip() == '[99]' else (row['policies_file_path'].split('/data/policies/')[-1].replace('.json', '') if isinstance(row['policies_file_path'], str) and '/data/policies/' in row['policies_file_path'] else None), axis=1)
```

```python
all_results_last_step_red.tail()
```

```python
# clean flood value columns
all_results_last_step_red['flood'] = all_results_last_step_red.apply(lambda row: 0 if (str(row['flood_time']).strip() == '[99]' and str(row['flood_depth_in_m']).strip() == '[99, 1, 1, 1, 1]') else (1 if str(row['flood_time']).strip() == '[10]' and str(row['flood_depth_in_m']).strip() == '[99, 1, 1, 1, 1]' else None), axis=1)
all_results_last_step_red = all_results_last_step_red[all_results_last_step_red['flood'].notna()]
all_results_last_step_red.head()
```

```python
# remove columns not needed anymore
all_results_last_step_red = all_results_last_step_red.drop(columns=['flood_time', 'flood_depth_in_m', 'policies_file_path', 'policy_activation'])
all_results_last_step_red.head()
```

```python
all_results_last_step_red.shape
```

```python
# share of adapted at start empirical - make a number (for sorting later)
all_results_last_step_red['set_hh_adapted_at_start'] = all_results_last_step_red['set_hh_adapted_at_start'].replace('False', 0.12)
all_results_last_step_red.head()
```

### Grouping results according to experiments

```python
all_results_last_step_red_grouped = all_results_last_step_red.groupby(['policy', 'flood', 'set_hh_adapted_at_start']).agg(['mean', 'std'])
```

```python
all_results_last_step_red_grouped
```

```python
# save results to csv
all_results_last_step_red_grouped.to_csv('../reports/tables/end_values_adaptation_damage_red.csv')
```
